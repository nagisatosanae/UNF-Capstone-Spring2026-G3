```python
# ============================================================
# 01. File paths
# ============================================================

base_path = "/Volumes/workspace/default/Yelp"

business_path = f"{base_path}/yelp_academic_dataset_business.json"
review_path   = f"{base_path}/yelp_academic_dataset_review.json"
checkin_path  = f"{base_path}/yelp_academic_dataset_checkin.json"

print(business_path)
print(review_path)
print(checkin_path)
```

```python
# ============================================================
# 02. Load raw JSON files
# ============================================================

business_df = spark.read.json(business_path)
review_df   = spark.read.json(review_path)
checkin_df  = spark.read.json(checkin_path)
```

```python
# ============================================================
# 03. Schema check
# ============================================================

business_df.printSchema()
review_df.printSchema()
checkin_df.printSchema()
```

```python
# ============================================================
# 04. Preview
# ============================================================

display(business_df.limit(5))
display(review_df.limit(5))
display(checkin_df.limit(5))
```

```python
# ============================================================
# 05. Clean business data
# ============================================================

from pyspark.sql.functions import col, lower, trim

business_clean = business_df.select(
    "business_id",
    "name",
    "city",
    "state",
    "latitude",
    "longitude",
    "stars",
    "review_count",
    "categories",
    "attributes",
    "hours",
    "is_open"
)

business_clean = business_clean.dropna(subset=["business_id", "categories"])
business_clean = business_clean.dropDuplicates(["business_id"])

business_clean = business_clean.filter(
    lower(col("categories")).contains("restaurant")
)

business_clean = business_clean.withColumn("city", trim(col("city"))) \
                               .withColumn("state", trim(col("state")))

display(business_clean.limit(10))
```

```python
# ============================================================
# 06. Clean review data
# ============================================================

from pyspark.sql.functions import trim

review_clean = review_df.select(
    "review_id",
    "business_id",
    "user_id",
    "stars",
    "text",
    "date",
    "useful",
    "funny",
    "cool"
)

review_clean = review_clean.dropna(
    subset=["review_id", "business_id", "stars", "text", "date"]
)

review_clean = review_clean.dropDuplicates(["review_id"])

review_clean = review_clean.filter(
    trim(col("text")) != ""
)

# Keep only restaurant-related reviews
review_clean = review_clean.join(
    business_clean.select("business_id"),
    on="business_id",
    how="inner"
)

display(review_clean.limit(10))
```

```python
# ============================================================
# 07. Clean check-in data
# ============================================================

checkin_clean = checkin_df.select(
    "business_id",
    "date"
)

checkin_clean = checkin_clean.dropna(subset=["business_id", "date"])
checkin_clean = checkin_clean.dropDuplicates(["business_id"])

# Keep only restaurant-related check-ins
checkin_clean = checkin_clean.join(
    business_clean.select("business_id"),
    on="business_id",
    how="inner"
)

display(checkin_clean.limit(10))
```

```python
# ============================================================
# 08. Row count check
# ============================================================

print("business_clean rows:", business_clean.count())
print("review_clean rows:", review_clean.count())
print("checkin_clean rows:", checkin_clean.count())

print("unique business in business_clean:", business_clean.select("business_id").distinct().count())
print("unique business in review_clean:", review_clean.select("business_id").distinct().count())
print("unique business in checkin_clean:", checkin_clean.select("business_id").distinct().count())
```

```python
# ============================================================
# 09. Save cleaned data as Parquet
# ============================================================

clean_base_path = "/Volumes/workspace/default/Yelp/cleaned"

business_clean.write.mode("overwrite").parquet(f"{clean_base_path}/business_clean")
review_clean.write.mode("overwrite").parquet(f"{clean_base_path}/review_clean")
checkin_clean.write.mode("overwrite").parquet(f"{clean_base_path}/checkin_clean")

print("Cleaned files saved to:", clean_base_path)
```

```python
# ============================================================
# 10. Verify saved parquet files
# ============================================================

business_clean_check = spark.read.parquet(f"{clean_base_path}/business_clean")
review_clean_check   = spark.read.parquet(f"{clean_base_path}/review_clean")
checkin_clean_check  = spark.read.parquet(f"{clean_base_path}/checkin_clean")

display(business_clean_check.limit(5))
display(review_clean_check.limit(5))
display(checkin_clean_check.limit(5))
```
